```python
# Temp-humidity-pressure relationships in a single reading

# Average across the two daily readings, gives one representative value per day per variable
core_df['temp_avg'] = core_df[['temp_9am', 'temp_3pm']].mean(axis=1)
core_df['humidity_avg'] = core_df[['humidity_9am', 'humidity_3pm']].mean(axis=1)
core_df['pressure_avg'] = core_df[['pressure_9am', 'pressure_3pm']].mean(axis=1)

# Daily swing — how much each variable moved from morning to afternoon
core_df['temp_swing'] = core_df['temp_3pm'] - core_df['temp_9am']
core_df['humidity_swing'] = core_df['humidity_3pm'] - core_df['humidity_9am']
core_df['pressure_swing'] = core_df['pressure_3pm'] - core_df['pressure_9am']

# Physical relationship: high temp usually pairs with lower humidity, and vice versa.
# This ratio breaks down (goes unusually high or low) when the relationship is violated.
core_df['temp_humidity_ratio'] = core_df['temp_avg'] / (core_df['humidity_avg'] + 1e-6)

# Range within the day
core_df['temp_range'] = core_df['temp_max'] - core_df['temp_min']

print("New shape:", core_df.shape)
print('station_id' in core_df.columns)
core_df[['station_id', 'date', 'temp_avg', 'humidity_avg', 'pressure_avg',
         'temp_swing', 'humidity_swing', 'pressure_swing', 'temp_humidity_ratio', 'temp_range']].head()
```

```python
from itertools import combinations

# One row per station with its coordinates
station_coords = core_df[['station_id', 'lat', 'lon']].drop_duplicates().reset_index(drop=True)
print("Stations:", len(station_coords))

# Haversine distance in km between two lat/lon points
def haversine(lat1, lon1, lat2, lon2):
    R = 6371  # Earth radius in km
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

# Build a station x station distance matrix
station_ids = station_coords['station_id'].values
n = len(station_ids)
dist_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i != j:
            dist_matrix[i, j] = haversine(
                station_coords.loc[i, 'lat'], station_coords.loc[i, 'lon'],
                station_coords.loc[j, 'lat'], station_coords.loc[j, 'lon']
            )

dist_df = pd.DataFrame(dist_matrix, index=station_ids, columns=station_ids)
print(dist_df.shape)
dist_df.head()
```

```python
K = 3  # number of nearby stations to compare against

nearest_neighbors = {}
for sid in station_ids:
    neighbors = dist_df.loc[sid].sort_values().iloc[1:K+1]  # skip index 0, that's itself (distance 0)
    nearest_neighbors[sid] = neighbors.index.tolist()

# Sanity check a couple of stations
for sid in list(nearest_neighbors.keys())[:5]:
    print(sid, "->", nearest_neighbors[sid])
```

```python
# Build a lookup: (station_id, date) -> row, for fast access
core_df = core_df.sort_values(['station_id', 'date']).reset_index(drop=True)

spatial_cols = ['temp_avg', 'humidity_avg', 'pressure_avg']

# Pivot each variable so we can look up any station's value on any date quickly
pivots = {col: core_df.pivot(index='date', columns='station_id', values=col) for col in spatial_cols}

# For each station, compute the neighbor average time series and the deviation
for col in spatial_cols:
    pivot = pivots[col]
    neighbor_avg_col = f'{col}_neighbor_avg'
    core_df[neighbor_avg_col] = np.nan

    for sid in station_ids:
        neighbors = nearest_neighbors[sid]
        # average of the neighbor stations' values, for every date
        neighbor_series = pivot[neighbors].mean(axis=1)
        # map this neighbor average onto the rows belonging to this station
        mask = core_df['station_id'] == sid
        core_df.loc[mask, neighbor_avg_col] = core_df.loc[mask, 'date'].map(neighbor_series)

    # deviation = how far this station is from its neighbors' average
    core_df[f'{col}_spatial_dev'] = core_df[col] - core_df[neighbor_avg_col]

print("New shape:", core_df.shape)
print('station_id' in core_df.columns)
core_df[['station_id', 'date', 'temp_avg', 'temp_avg_neighbor_avg', 'temp_avg_spatial_dev']].head(10)
```

```python
def add_temporal_features(df, value_cols):
    df = df.sort_values(['station_id', 'date']).reset_index(drop=True)
    grouped = df.groupby('station_id')
    for col in value_cols:
        df[f'{col}_roll_mean7'] = grouped[col].transform(lambda s: s.rolling(7, min_periods=3).mean())
        df[f'{col}_roll_std7'] = grouped[col].transform(lambda s: s.rolling(7, min_periods=3).std())
        df[f'{col}_diff1'] = grouped[col].transform(lambda s: s.diff())
    for col in value_cols:
        df[f'{col}_zscore'] = (df[col] - df[f'{col}_roll_mean7']) / df[f'{col}_roll_std7']
    return df

def add_multivariate_features(df):
    df['temp_avg'] = df[['temp_9am', 'temp_3pm']].mean(axis=1)
    df['humidity_avg'] = df[['humidity_9am', 'humidity_3pm']].mean(axis=1)
    df['pressure_avg'] = df[['pressure_9am', 'pressure_3pm']].mean(axis=1)
    df['temp_swing'] = df['temp_3pm'] - df['temp_9am']
    df['humidity_swing'] = df['humidity_3pm'] - df['humidity_9am']
    df['pressure_swing'] = df['pressure_3pm'] - df['pressure_9am']
    df['temp_humidity_ratio'] = df['temp_avg'] / (df['humidity_avg'] + 1e-6)
    df['temp_range'] = df['temp_max'] - df['temp_min']
    return df

def add_spatial_features(df, nearest_neighbors, station_ids, spatial_cols=['temp_avg', 'humidity_avg', 'pressure_avg']):
    df = df.sort_values(['station_id', 'date']).reset_index(drop=True)
    pivots = {col: df.pivot(index='date', columns='station_id', values=col) for col in spatial_cols}
    for col in spatial_cols:
        pivot = pivots[col]
        neighbor_avg_col = f'{col}_neighbor_avg'
        df[neighbor_avg_col] = np.nan
        for sid in station_ids:
            neighbors = nearest_neighbors[sid]
            neighbor_series = pivot[neighbors].mean(axis=1)
            mask = df['station_id'] == sid
            df.loc[mask, neighbor_avg_col] = df.loc[mask, 'date'].map(neighbor_series)
        df[f'{col}_spatial_dev'] = df[col] - df[neighbor_avg_col]
    return df

print("Functions defined")
```

```python
def inject_anomalies(df, value_cols, fraction=0.04, seed=42):
    df = df.copy().reset_index(drop=True)
    rng = np.random.default_rng(seed)
    df['is_anomaly'] = 0
    df['anomaly_type'] = 'normal'

    n_inject = int(len(df) * fraction)
    inject_idx = rng.choice(df.index, size=n_inject, replace=False)
    types = ['spike', 'jump', 'frozen', 'drift']

    for idx in inject_idx:
        if df.at[idx, 'is_anomaly'] == 1:
            continue  # already touched by an earlier injection
        col = rng.choice(value_cols)
        atype = rng.choice(types)
        sid = df.at[idx, 'station_id']

        if atype == 'spike':
            df.at[idx, col] += rng.choice([-1, 1]) * rng.uniform(15, 30)
            df.at[idx, 'is_anomaly'] = 1
            df.at[idx, 'anomaly_type'] = 'spike'

        elif atype == 'jump':
            df.at[idx, col] += rng.choice([-1, 1]) * rng.uniform(8, 15)
            df.at[idx, 'is_anomaly'] = 1
            df.at[idx, 'anomaly_type'] = 'jump'

        elif atype == 'frozen':
            station_rows = df[df['station_id'] == sid].sort_values('date')
            pos = list(station_rows.index).index(idx)
            freeze_len = rng.integers(3, 7)
            freeze_idx = station_rows.index[pos:pos + freeze_len]
            frozen_value = df.at[idx, col]
            df.loc[freeze_idx, col] = frozen_value
            df.loc[freeze_idx, 'is_anomaly'] = 1
            df.loc[freeze_idx, 'anomaly_type'] = 'frozen'

        elif atype == 'drift':
            station_rows = df[df['station_id'] == sid].sort_values('date')
            pos = list(station_rows.index).index(idx)
            drift_len = rng.integers(5, 10)
            drift_idx = station_rows.index[pos:pos + drift_len]
            drift_amounts = np.linspace(0, rng.uniform(8, 15), len(drift_idx))
            df.loc[drift_idx, col] = df.loc[drift_idx, col].values + drift_amounts
            df.loc[drift_idx, 'is_anomaly'] = 1
            df.loc[drift_idx, 'anomaly_type'] = 'drift'

    return df

# Reload the RAW clean data (before any features) and inject into that
raw_clean = pd.read_csv('data/processed/skyguard_clean.csv')
raw_clean['date'] = pd.to_datetime(raw_clean['date'])

value_cols = ['temp_9am', 'temp_3pm', 'temp_min', 'temp_max',
              'humidity_9am', 'humidity_3pm', 'pressure_9am', 'pressure_3pm']

injected_df = inject_anomalies(raw_clean, value_cols, fraction=0.04, seed=42)

print("Total rows:", len(injected_df))
print("\nAnomaly type counts:")
print(injected_df['anomaly_type'].value_counts())
```

```python
final_df = add_temporal_features(injected_df, value_cols)
final_df = add_multivariate_features(final_df)
final_df = add_spatial_features(final_df, nearest_neighbors, station_ids)

print("Final shape:", final_df.shape)
print('is_anomaly' in final_df.columns, 'station_id' in final_df.columns)
final_df[final_df['is_anomaly'] == 1][['station_id', 'date', 'anomaly_type', 'temp_9am', 'temp_9am_zscore']].head(10)
```

```python
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import StandardScaler

model_df = pd.read_csv('data/processed/skyguard_labeled.csv')
model_df['date'] = pd.to_datetime(model_df['date'])

# Feature columns: everything engineered, excluding identifiers and raw labels
exclude_cols = ['station_id', 'date', 'lat', 'lon', 'is_anomaly', 'anomaly_type']
feature_cols = [c for c in model_df.columns if c not in exclude_cols]
print("Number of features:", len(feature_cols))
print(feature_cols)

# Drop rows where features couldn't be computed (start of each station's series)
model_df_clean = model_df.dropna(subset=feature_cols).reset_index(drop=True)
print("\nRows before dropna:", len(model_df), " after:", len(model_df_clean))
print("Anomaly rate after dropna:", model_df_clean['is_anomaly'].mean())
```

```python
model_df_clean = model_df_clean.sort_values('date').reset_index(drop=True)

split_date = model_df_clean['date'].quantile(0.8)

train_df = model_df_clean[model_df_clean['date'] < split_date]
test_df = model_df_clean[model_df_clean['date'] >= split_date]

print("Train:", train_df.shape, "| Test:", test_df.shape)

print("Train date range:", train_df['date'].min(), "-", train_df['date'].max())
print("Test date range:", test_df['date'].min(), "-", test_df['date'].max())

print("\nTrain anomaly rate:", train_df['is_anomaly'].mean())
print("Test anomaly rate:", test_df['is_anomaly'].mean())
```

```python
X_train = train_df[feature_cols]
X_test = test_df[feature_cols]
y_test = test_df['is_anomaly']  # ground truth, only used for evaluation, never for training

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# contamination = expected proportion of anomalies, matches our injected rate
model = IsolationForest(
    n_estimators=200,
    contamination=0.123,
    max_samples='auto',
    random_state=42,
    n_jobs=-1
)
model.fit(X_train_scaled)

print("Model trained")
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Isolation Forest outputs -1 for anomaly, 1 for normal — convert to our 0/1 convention
raw_preds = model.predict(X_test_scaled)
y_pred = np.where(raw_preds == -1, 1, 0)

# Anomaly score: lower = more anomalous. Flip sign so higher = more anomalous (more intuitive)
anomaly_scores = -model.decision_function(X_test_scaled)

precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print(f"Precision: {precision:.3f}")
print(f"Recall:    {recall:.3f}")
print(f"F1-score:  {f1:.3f}")

print("\nConfusion matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nFull report:")
print(classification_report(y_test, y_pred, target_names=['normal', 'anomaly']))
```

```python
test_df_eval = test_df.copy()
test_df_eval['predicted'] = y_pred
test_df_eval['anomaly_score'] = anomaly_scores

detection_by_type = test_df_eval[test_df_eval['is_anomaly'] == 1].groupby('anomaly_type')['predicted'].mean()
print("Detection rate by anomaly type:")
print(detection_by_type.sort_values(ascending=False))
```

```python
def add_temporal_features_v2(df, value_cols):
    df = df.sort_values(['station_id', 'date']).reset_index(drop=True)
    grouped = df.groupby('station_id')
    for col in value_cols:
        # shift(1) BEFORE rolling — baseline uses only the 7 days strictly before today
        df[f'{col}_roll_mean7'] = grouped[col].transform(lambda s: s.shift(1).rolling(7, min_periods=3).mean())
        df[f'{col}_roll_std7']  = grouped[col].transform(lambda s: s.shift(1).rolling(7, min_periods=3).std())
        df[f'{col}_roll_mean14']= grouped[col].transform(lambda s: s.shift(1).rolling(14, min_periods=5).mean())
        df[f'{col}_roll_std3']  = grouped[col].transform(lambda s: s.shift(1).rolling(3, min_periods=2).std())
        df[f'{col}_diff1']      = grouped[col].transform(lambda s: s.diff())
    for col in value_cols:
        df[f'{col}_zscore']   = (df[col] - df[f'{col}_roll_mean7']) / df[f'{col}_roll_std7']
        df[f'{col}_mean_gap'] = df[f'{col}_roll_mean7'] - df[f'{col}_roll_mean14']
    for col in value_cols:
        same = grouped[col].transform(lambda s: s.diff() == 0)
        df[f'{col}_unchanged_streak'] = same.groupby(df['station_id']).transform(
            lambda s: s.groupby((~s).cumsum()).cumcount()
        )
    return df

def add_spatial_features_v2(df, nearest_neighbors, station_ids, spatial_cols=['temp_avg','humidity_avg','pressure_avg']):
    df = df.sort_values(['station_id', 'date']).reset_index(drop=True)
    pivots = {col: df.pivot(index='date', columns='station_id', values=col) for col in spatial_cols}
    for col in spatial_cols:
        pivot = pivots[col]
        df[f'{col}_neighbor_avg'] = np.nan
        for sid in station_ids:
            neighbor_series = pivot[nearest_neighbors[sid]].mean(axis=1)
            mask = df['station_id'] == sid
            df.loc[mask, f'{col}_neighbor_avg'] = df.loc[mask, 'date'].map(neighbor_series)
        df[f'{col}_spatial_dev'] = df[col] - df[f'{col}_neighbor_avg']
        # normalize AWAY each station's permanent offset from neighbors: z-score of the deviation itself, lagged
        grouped_dev = df.groupby('station_id')[f'{col}_spatial_dev']
        dev_mean = grouped_dev.transform(lambda s: s.shift(1).rolling(14, min_periods=5).mean())
        dev_std  = grouped_dev.transform(lambda s: s.shift(1).rolling(14, min_periods=5).std())
        df[f'{col}_spatial_dev_zscore'] = (df[f'{col}_spatial_dev'] - dev_mean) / dev_std
    return df

model_df = pd.read_csv('data/processed/skyguard_labeled.csv')
model_df['date'] = pd.to_datetime(model_df['date'])

value_cols = ['temp_9am', 'temp_3pm', 'temp_min', 'temp_max',
              'humidity_9am', 'humidity_3pm', 'pressure_9am', 'pressure_3pm']

model_df = add_temporal_features_v2(model_df, value_cols)
model_df = add_multivariate_features(model_df)
model_df = add_spatial_features_v2(model_df, nearest_neighbors, station_ids)
model_df = model_df.replace([np.inf, -np.inf], np.nan)

print("Shape:", model_df.shape)
```

```python
def inject_anomalies_v2(df, value_cols, fraction=0.04, seed=42):
    df = df.copy().reset_index(drop=True)
    rng = np.random.default_rng(seed)
    df['is_anomaly'] = 0
    df['anomaly_type'] = 'normal'
    df['anomaly_col'] = ''  # NEW: track which column was actually corrupted

    n_inject = int(len(df) * fraction)
    inject_idx = rng.choice(df.index, size=n_inject, replace=False)
    types = ['spike', 'jump', 'frozen', 'drift']

    for idx in inject_idx:
        if df.at[idx, 'is_anomaly'] == 1:
            continue
        col = rng.choice(value_cols)
        atype = rng.choice(types)
        sid = df.at[idx, 'station_id']

        if atype == 'spike':
            df.at[idx, col] += rng.choice([-1, 1]) * rng.uniform(15, 30)
            df.at[idx, 'is_anomaly'] = 1
            df.at[idx, 'anomaly_type'] = 'spike'
            df.at[idx, 'anomaly_col'] = col

        elif atype == 'jump':
            df.at[idx, col] += rng.choice([-1, 1]) * rng.uniform(8, 15)
            df.at[idx, 'is_anomaly'] = 1
            df.at[idx, 'anomaly_type'] = 'jump'
            df.at[idx, 'anomaly_col'] = col

        elif atype == 'frozen':
            station_rows = df[df['station_id'] == sid].sort_values('date')
            pos = list(station_rows.index).index(idx)
            freeze_len = rng.integers(3, 7)
            freeze_idx = station_rows.index[pos:pos + freeze_len]
            frozen_value = df.at[idx, col]
            df.loc[freeze_idx, col] = frozen_value
            df.loc[freeze_idx, 'is_anomaly'] = 1
            df.loc[freeze_idx, 'anomaly_type'] = 'frozen'
            df.loc[freeze_idx, 'anomaly_col'] = col

        elif atype == 'drift':
            station_rows = df[df['station_id'] == sid].sort_values('date')
            pos = list(station_rows.index).index(idx)
            drift_len = rng.integers(5, 10)
            drift_idx = station_rows.index[pos:pos + drift_len]
            drift_amounts = np.linspace(0, rng.uniform(8, 15), len(drift_idx))
            df.loc[drift_idx, col] = df.loc[drift_idx, col].values + drift_amounts
            df.loc[drift_idx, 'is_anomaly'] = 1
            df.loc[drift_idx, 'anomaly_type'] = 'drift'
            df.loc[drift_idx, 'anomaly_col'] = col

    return df

raw_clean = pd.read_csv('data/processed/skyguard_clean.csv')
raw_clean['date'] = pd.to_datetime(raw_clean['date'])

injected_df2 = inject_anomalies_v2(raw_clean, value_cols, fraction=0.04, seed=42)

model_df2 = add_temporal_features_v2(injected_df2, value_cols)
model_df2 = add_multivariate_features(model_df2)
model_df2 = add_spatial_features_v2(model_df2, nearest_neighbors, station_ids)
model_df2 = model_df2.replace([np.inf, -np.inf], np.nan)

print("Shape:", model_df2.shape)
print(model_df2['anomaly_col'].value_counts())
```

```python
def explain_alert(row_index, X_data, shap_vals, feature_names, top_n=5):

    row_shap = shap_vals[row_index]
    actual_values = X_data.iloc[row_index].values

    contributions = pd.DataFrame({
        'feature': feature_names,
        'actual_value': actual_values,
        'shap_value': row_shap
    })

    # Features pushing toward anomaly
    anomaly_features = (
        contributions[contributions['shap_value'] > 0]
        .sort_values('shap_value', ascending=False)
        .head(top_n)
    )

    # Features pushing toward normal
    normal_features = (
        contributions[contributions['shap_value'] < 0]
        .sort_values('shap_value', ascending=True)
        .head(top_n)
    )

    return anomaly_features, normal_features
```

```python
import asyncio, websockets, json

async def test_stream():
    uri = "ws://localhost:8000/ws"
    async with websockets.connect(uri) as ws:
        for _ in range(10):
            message = await ws.recv()
            data = json.loads(message)
            flag = "🔴 ANOMALY" if data["is_anomaly"] else "🟢 normal"
            print(f"{flag} | {data['station_id']} {data['date']} | T={data['temperature']}°C H={data['humidity']}% P={data['pressure']}hPa | prob={data['probability']}")
            if data["is_anomaly"]:
                for reason in data["reasons"]:
                    print("   ✓", reason)

await test_stream()
```

```python
import asyncio, websockets, json

async def test_stream_longer():
    uri = "ws://localhost:8000/ws"
    async with websockets.connect(uri) as ws:
        for _ in range(60):  # ~60 seconds, more likely to catch a real anomaly
            message = await ws.recv()
            data = json.loads(message)
            if data["is_anomaly"]:
                print(f"🔴 ANOMALY | {data['station_id']} {data['date']} | prob={data['probability']}")
                for reason in data["reasons"]:
                    print("   ✓", reason)

await test_stream_longer()
```
